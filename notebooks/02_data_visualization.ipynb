### KPI_Metrics

```python
class KPI_Metrics:
    total_rows = data_frame.shape[0]
    total_columns = data_frame.shape[1]

    total_days = len(data_frame["order_date"].unique())
    total_orders = len(data_frame["order_id"].unique())
    total_customers = len(data_frame["customer_id"].unique())
    total_products = len(data_frame["product_id"].unique())
    total_quantity_sold = data_frame["quantity"].sum()
    total_countries = len(data_frame["country"].unique())
    total_sales = data_frame["sales"].sum()
    total_profit = data_frame["profit"].sum()
    total_shipping_cost = data_frame["shipping_cost"].sum()
    total_cost = data_frame["cost"].sum()

    avg_orders_per_customer = round(total_orders / total_customers, 2)
    avg_products_per_customer = round(total_quantity_sold / total_customers, 2)
    avg_revenue_per_customer = round(data_frame["sales"].sum() / total_customers, 2)
    avg_profit_per_customer = round(data_frame["profit"].sum() / total_customers, 2)

    avg_products_per_order = round(total_quantity_sold / total_orders, 2)
    avg_unique_products_per_order = round(data_frame.groupby("order_id")["product_id"].nunique().mean(), 2)
    avg_revenue_per_order = round(data_frame["sales"].sum() / total_orders, 2)
    avg_profit_per_order = round(data_frame["profit"].sum() / total_orders, 2)

    profit_margin = round((data_frame["profit"].sum() / data_frame["sales"].sum()) * 100, 2)
    return_rate = round(data_frame["returned"].mean() * 100, 2)
    total_return = len(data_frame.loc[data_frame["returned"] == True]["order_id"].unique())
    profit_percentage = round((data_frame.loc[data_frame["profit"] > 0]["order_id"].count() / total_rows) * 100, 2)
    loss_percentage = round((data_frame.loc[data_frame["profit"] <= 0]["order_id"].count() / total_rows) * 100, 2)

    @staticmethod
    def compute_metrics(filtered_df):
        KPI_Metrics.total_rows = filtered_df.shape[0]
        KPI_Metrics.total_columns = filtered_df.shape[1]

        KPI_Metrics.total_days = len(filtered_df["order_date"].unique())
        KPI_Metrics.total_orders = len(filtered_df["order_id"].unique())
        KPI_Metrics.total_customers = len(filtered_df["customer_id"].unique())
        KPI_Metrics.total_products = len(filtered_df["product_id"].unique())
        KPI_Metrics.total_quantity_sold = filtered_df["quantity"].sum()
        KPI_Metrics.total_countries = len(filtered_df["country"].unique())
        KPI_Metrics.total_sales = filtered_df["sales"].sum()
        KPI_Metrics.total_profit = filtered_df["profit"].sum()
        KPI_Metrics.total_shipping_cost = filtered_df["shipping_cost"].sum()
        KPI_Metrics.total_cost = filtered_df["cost"].sum()

        KPI_Metrics.avg_orders_per_customer = round(KPI_Metrics.total_orders / KPI_Metrics.total_customers, 2)
        KPI_Metrics.avg_products_per_customer = round(KPI_Metrics.total_quantity_sold / KPI_Metrics.total_customers, 2)
        KPI_Metrics.avg_revenue_per_customer = round(filtered_df["sales"].sum() / KPI_Metrics.total_customers, 2)
        KPI_Metrics.avg_profit_per_customer = round(filtered_df["profit"].sum() / KPI_Metrics.total_customers, 2)

        KPI_Metrics.avg_products_per_order = round(KPI_Metrics.total_quantity_sold / KPI_Metrics.total_orders, 2)
        KPI_Metrics.avg_unique_products_per_order = round(filtered_df.groupby("order_id")["product_id"].nunique().mean(), 2)
        KPI_Metrics.avg_revenue_per_order = round(filtered_df["sales"].sum() / KPI_Metrics.total_orders, 2)
        KPI_Metrics.avg_profit_per_order = round(filtered_df["profit"].sum() / KPI_Metrics.total_orders, 2)

        KPI_Metrics.profit_margin = round((filtered_df["profit"].sum() / filtered_df["sales"].sum()) * 100, 2)
        KPI_Metrics.return_rate = round(filtered_df["returned"].mean() * 100, 2)
        KPI_Metrics.total_return = len(filtered_df.loc[filtered_df["returned"] == True]["order_id"].unique())
        KPI_Metrics.profit_percentage = round((filtered_df.loc[filtered_df["profit"] > 0]["order_id"].count() / KPI_Metrics.total_rows) * 100, 2)
        KPI_Metrics.loss_percentage = round((filtered_df.loc[filtered_df["profit"] <= 0]["order_id"].count() / KPI_Metrics.total_rows) * 100, 2)
```

```python
line_chart_color_palette  = [
    "#F0EBD8",
    "#E07E50",
    "#0D1321", 
    "#1D2D44",  
    "#3E5C76", 
    "#FDF0D5", 
    "#5A748C", 
    "#8DA2B8",
    "#A7B7C9",
    "#C2A97A",   
    "#C0CCD8",  
    "#C96A42"  
]

metric_config = {
    "delivery_time": {"agg": "mean", "format": ".2f"},
    "quantity": {"agg": "sum", "format": ",.0f"},
    "discount": {"agg": "mean", "format": ".2%"},
    "sales": {"agg": "sum", "format": ",.0f"},
    "profit": {"agg": "sum", "format": ",.0f"},
    "profit_margin": {"agg": "mean", "format": ".2%"},
    "cost": {"agg": "sum", "format": ",.0f"},
    "shipping_cost": {"agg": "sum", "format": ",.0f"},
    "shipping_ratio": {"agg": "mean", "format": ".2%"},
    "cost_ratio": {"agg": "mean", "format": ".2%"}
}

metric_display_names = {
    "delivery_time": "Delivery Time",
    "quantity": "Quantity",
    "discount": "Discount",
    "sales": "Sales",
    "profit": "Profit",
    "profit_margin": "Profit Margin",
    "cost": "Cost",
    "shipping_cost": "Shipping Cost",
    "shipping_ratio": "Shipping Ratio",
    "cost_ratio": "Cost Ratio"
}

time_granularity_labels = {
    "month": "Monthly",
    "quarter": "Quarterly",
    "year": "Yearly"
}

dimension_labels = {
    "None": "",
    "category": "by Category",
    "segment": "by Segment",
    "order_priority": "by Order Priority",
    "ship_mode": "by Ship Mode",
    "market": "by Market",
    "sub_category": "by Sub-Category",
    "country": "by Country",
    "customer_name": "by Customer"
}

scatterplot_color_palette  = [
    "#F0EBD8",
    "#E07E50",
    "#0D1321", 
    "#1D2D44",  
    "#3E5C76", 
    "#FDF0D5", 
    "#5A748C", 
    "#8DA2B8",
    "#A7B7C9",
    "#C2A97A",   
    "#C0CCD8",  
    "#C96A42"  
]





def build_kpi_cards():
    kpi_cards_data = [
    {
        "title": "Total revenue",
        "text": f"${int(KPI_Metrics.total_sales): ,}"
    },
    {
        "title": "Total quantity sold",
        "text":  f"{int(KPI_Metrics.total_quantity_sold): ,}"
    },
    {
        "title": "Profit",
        "text": f"${int(KPI_Metrics.total_profit): ,}"
    },
    {
        "title": "Profit margin",
        "text": f"{KPI_Metrics.profit_margin}%"
    },
    {
        "title": "Orders count",
        "text": f"{int(KPI_Metrics.total_orders): ,}"
    },
    {
        "title": "Returns",
        "text":  f"{int(KPI_Metrics.total_return): ,}"
    },
    {
        "title": "Returns rate",
        "text":  f"{KPI_Metrics.return_rate}%"
    },
    {
        "title": "Total products",
        "text":  f"{int(KPI_Metrics.total_products): ,}"
    },
    {
        "title": "Customers",
        "text":  f"{int(KPI_Metrics.total_customers): ,}"
    },
    {
        "title": "Countries",
        "text":  f"{KPI_Metrics.total_countries}"
    },
    {
        "title": "Shipping cost",
        "text":  f"${int(KPI_Metrics.total_shipping_cost): ,}"
    },
    {
        "title": "Total cost",
        "text":  f"${int(KPI_Metrics.total_cost): ,}"
    },
    ]


    kpi_cards = []

    for kpi_card_data in kpi_cards_data:
        kpi_card_text_div = html.Div(kpi_card_data["text"], style = kpi_card_text_style)
        kpi_card_title_div = html.Div(kpi_card_data["title"], style = kpi_card_title_style)
        kpi_card = html.Div([kpi_card_title_div, kpi_card_text_div], style = kpi_card_style)
        kpi_cards.append(kpi_card)

    return kpi_cards

def build_customer_and_order():
    customer_info = html.Div([
        html.H3("Customer info:", style = basic_stats_text_style),
        html.Ul([
            html.Li(f"Average orders per customer: {KPI_Metrics.avg_orders_per_customer}"),
            html.Li(f"Average products per customer: {KPI_Metrics.avg_products_per_customer}"),
            html.Li(f"Average revenue per customer: {KPI_Metrics.avg_revenue_per_customer}"),
            html.Li(f"Average profit per customer: {KPI_Metrics.avg_profit_per_customer}")
        ], style = basic_stats_text_style)
        ]
    )

    order_info = html.Div([
        html.H3("Order info:", style = basic_stats_text_style),
        html.Ul([
            html.Li(f"Average products per order: {KPI_Metrics.avg_products_per_order}"),
            html.Li(f"Average unique products per order: {KPI_Metrics.avg_unique_products_per_order}"),
            html.Li(f"Average revenue per order: {KPI_Metrics.avg_revenue_per_order}"),
            html.Li(f"Average profit per order: {KPI_Metrics.avg_profit_per_order}")
        ], style = basic_stats_text_style)
        ]
    )

    return customer_info, order_info

def build_profit_distribution():
    profit_loss_distribution_df = pd.DataFrame({
        "Profit Type": ["Profit", "Loss"], 
        "Percentage of Sales (%)": [KPI_Metrics.profit_percentage, KPI_Metrics.loss_percentage]
    })
    profit_loss_distribution_fig = px.bar(data_frame = profit_loss_distribution_df, x = "Profit Type", y = "Percentage of Sales (%)", width = 350, height = 350)
    profit_loss_distribution_fig.update_yaxes(fixedrange = True, range = [0, 100], dtick = 10, tickfont_color = "#0D1321", tickfont_family = "Arial", tickfont_weight = 600, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", gridcolor = "#F0EBD8", zeroline = False, ticklabelstandoff = 5)
    profit_loss_distribution_fig.update_xaxes(fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", tickfont_weight = 600, tickfont_family = "Arial",  tickfont_color = "#0D1321", ticklabelstandoff = 5)
    profit_loss_distribution_fig.update_traces(width = 0.5, marker_color = "#3E5C76", marker_line_color = "#1D2D44", marker_line_width = 1.5, opacity = 1)
    profit_loss_distribution_fig.update_layout(margin=dict(
            l=25,
            r=25,
            b=25,
            t=50,
            pad=5),
            paper_bgcolor="#FDF0D5",
            plot_bgcolor = "#748CAB",   
            title_text = "Profit vs Loss Distribution",
            title_font_color = "#1D2D44",
            title_font_family = "Arial",
            title_font_size = 18,
            title_font_weight = 600,
            title_y = 0.96)
    
    return profit_loss_distribution_fig

def build_boxplot(filtered_df, metric):
    metric_box_plot_fig = px.box(data_frame = filtered_df, y = metric, width = 343, height = 350)
    metric_box_plot_fig.update_yaxes(fixedrange = True, tickfont_color = "#0D1321", tickfont_family = "Arial", tickfont_weight = 600, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", gridcolor = "#F0EBD8", zeroline = False, ticklabelstandoff = 5)
    metric_box_plot_fig.update_xaxes(fixedrange = True)
    metric_box_plot_fig.update_traces(boxmean = True, boxpoints = False, width = 0.5, marker_color = "#1D2D44", opacity = 1, fillcolor = "#3E5C76", line_width = 1.5)
    metric_box_plot_fig.update_layout(margin=dict(
            l=25,
            r=25,
            b=25,
            t=50,
            pad=5),
            paper_bgcolor="#FDF0D5",
            plot_bgcolor = "#748CAB",
            title_text = "Metric Overview",
            title_font_color = "#1D2D44",
            title_font_family = "Arial",
            title_font_size = 18,
            title_font_weight = 600,
            title_y = 0.96
    )

    return metric_box_plot_fig

def build_histogram(filtered_df, metric):
    metric_histogram_fig = px.histogram(data_frame = filtered_df, x = metric, width = 500, height = 350, nbins = metrics_distribution[metric])
    metric_histogram_fig.update_yaxes(fixedrange = True, tickfont_color = "#0D1321", tickfont_family = "Arial", tickfont_weight = 600, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", gridcolor = "#F0EBD8", zeroline = False, ticklabelstandoff = 5)
    metric_histogram_fig.update_xaxes(fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", tickfont_weight = 600, tickfont_family = "Arial",  tickfont_color = "#0D1321", ticklabelstandoff = 5)
    metric_histogram_fig.update_traces(marker_color = "#3E5C76", marker_line_color = "#1D2D44", marker_line_width = 1.5, opacity = 1)
    metric_histogram_fig.update_layout(margin=dict(
            l=25,
            r=25,
            b=25,
            t=50,
            pad=5),
            paper_bgcolor="#FDF0D5",
            plot_bgcolor = "#748CAB",
            title_text = "Metric Distribution",
            title_font_color = "#1D2D44",
            title_font_family = "Arial",
            title_font_size = 18,
            title_font_weight = 600,
            title_y = 0.96
    )

    return metric_histogram_fig

@app.callback([Output("kpi_cards_div", "children"), Output("customer_and_order_info_div", "children"),  Output("profit_loss_distribution_graph", "figure"), Output("metric_box_plot_graph", "figure"), Output("metric_histogram_graph", "figure")], 
              [Input("kpi_year_slider", "value"), Input("kpi_category_dropdown", "value"), Input("kpi_market_dropdown", "value"), Input("change_metric_dropdown", "value")])
def update_kpi_and_stats_dashboard(year_index, category, market, metric):
    filtered_df = data_frame
    if kpi_year_slider_marks[year_index] != "All":
        filtered_df = filtered_df.loc[data_frame["order_year"] == years[year_index]]
    if category != "All":
        filtered_df = filtered_df.loc[data_frame["category"] == category]
    if market != "All":
        filtered_df = filtered_df.loc[data_frame["market"] == market]

    KPI_Metrics.compute_metrics(filtered_df)

    if ctx.triggered_id == "change_metric_dropdown":
        return no_update, no_update, no_update, build_boxplot(filtered_df, metric), build_histogram(filtered_df, metric)
    else:
        return build_kpi_cards(), build_customer_and_order(), build_profit_distribution(), build_boxplot(filtered_df, metric), build_histogram(filtered_df, metric)
   




@app.callback(Output("line_chart_graph", "figure"), [Input("linechart_time_granularity_dropdown", "value"), Input("linechart_filter_dropdown", "value"), Input("linechart_metric_dropdown", "value")])
def update_linechart(time_granularity, dimension, metric):
    line_chart_df = data_frame.copy()
    line_chart_df["order_date"] = pd.to_datetime(line_chart_df["order_date"])

    if time_granularity == "month":
        line_chart_df["date"] = line_chart_df["order_date"].dt.to_period("M").dt.to_timestamp()
        line_chart_df["label"] = line_chart_df["date"].dt.strftime("%b %Y")

    elif time_granularity == "quarter":
        line_chart_df["date"] = line_chart_df["order_date"].dt.to_period("Q").dt.to_timestamp()
        line_chart_df["label"] = (
            line_chart_df["date"].dt.year.astype(str) +
            " Q" +
            line_chart_df["date"].dt.quarter.astype(str)
        )

    elif time_granularity == "year":
        line_chart_df["date"] = line_chart_df["order_date"].dt.to_period("Y").dt.to_timestamp()
        line_chart_df["label"] = line_chart_df["date"].dt.year.astype(str)


    group_keys = [dimension, "date", "label"] if dimension != "None" else ["date", "label"]
    if metric == "profit_margin":
        line_chart_df = line_chart_df.groupby(group_keys).agg({"profit": "sum", "sales": "sum"}).reset_index()
        line_chart_df["profit_margin"] = round(line_chart_df["profit"] / line_chart_df["sales"], 4)
    elif metric == "shipping_ratio":
        line_chart_df = line_chart_df.groupby(group_keys).agg({"shipping_cost": "sum", "sales": "sum"}).reset_index()
        line_chart_df["shipping_ratio"] = round(line_chart_df["shipping_cost"] / line_chart_df["sales"], 4)
    elif metric == "cost_ratio":
        line_chart_df = line_chart_df.groupby(group_keys).agg({"cost": "sum", "sales": "sum"}).reset_index()
        line_chart_df["cost_ratio"] = round(line_chart_df["cost"] / line_chart_df["sales"], 4)
    else:
        line_chart_df = line_chart_df.groupby(group_keys).agg({metric : metric_config[metric]["agg"]}).reset_index()

    line_chart_df = line_chart_df.sort_values(by="date")


    metric_name = metric_display_names[metric]
    time_name = time_granularity_labels[time_granularity]
    dimension_name = dimension_labels[dimension]

    title = f"{time_name} {metric_name}"
    if dimension_name:
        title += f" {dimension_name}"


    line_chart_fig = px.line(data_frame = line_chart_df, x = "date", y = metric, color = (None if dimension == "None" else dimension), color_discrete_sequence = line_chart_color_palette)
    line_chart_fig.update_layout(
                yaxis={"tickformat": ",.0f"}, 
                paper_bgcolor="#FDF0D5",
                plot_bgcolor = "#748CAB",   
                title_text = title,
                title_font_color = "#1D2D44",
                title_font_family = "Arial",
                title_font_size = 18,
                title_font_weight = 600,
                title_y = 0.96, 
                margin=dict(l=25, r=25, b=25, t=50, pad = 5))

    unique_dates_df = line_chart_df[["date", "label"]].drop_duplicates().sort_values(by="date")

    line_chart_fig.update_traces(mode="markers+lines", marker_line_width = 1.5, line_width = 2, marker_size = 5)
    line_chart_fig.update_xaxes(tickangle = 45, range = [unique_dates_df["date"].min(), unique_dates_df["date"].max()], tickvals = line_chart_df["date"], ticktext=line_chart_df["label"], showgrid = False, title_text = "Date", fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", tickfont_weight = 600, tickfont_family = "Arial", tickfont_size = 9,  tickfont_color = "#0D1321", ticklabelstandoff = 5)
    line_chart_fig.update_yaxes(autorange = True, tickformat =  metric_config[metric]["format"], zeroline = False, fixedrange = True, tickfont_color = "#0D1321", tickfont_family = "Arial", tickfont_weight = 600, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", gridcolor = "#F0EBD8", ticklabelstandoff = 5)    

    return line_chart_fig

is_barchart_sorted = True
@app.callback([Output("bar_chart_graph", "figure"), Output("barchart_sort_icon_arrow", "style")], [Input("barchart_dimension_dropdown", "value"), Input("barchart_metric_dropdown", "value"), Input("barchart_sort_button", "n_clicks")])
def update_barchart(dimension, metric, sorting_button_clicks):
    global is_barchart_sorted
    if ctx.triggered_id == "barchart_sort_button": 
        is_barchart_sorted = not is_barchart_sorted

    bar_chart_df = data_frame.copy()

    if metric == "profit_margin":
        bar_chart_df = bar_chart_df.groupby(dimension).agg({"profit": "sum", "sales": "sum"}).reset_index()
        bar_chart_df["profit_margin"] = round(bar_chart_df["profit"] / bar_chart_df["sales"], 4)
    elif metric == "shipping_ratio":
        bar_chart_df = bar_chart_df.groupby(dimension).agg({"shipping_cost": "sum", "sales": "sum"}).reset_index()
        bar_chart_df["shipping_ratio"] = round(bar_chart_df["shipping_cost"] / bar_chart_df["sales"], 4)
    elif metric == "cost_ratio":
        bar_chart_df = bar_chart_df.groupby(dimension).agg({"cost": "sum", "sales": "sum"}).reset_index()
        bar_chart_df["cost_ratio"] = round(bar_chart_df["cost"] / bar_chart_df["sales"], 4)
    else:
        bar_chart_df = bar_chart_df.groupby(dimension).agg({metric : metric_config[metric]["agg"]}).reset_index()


    bar_chart_df = bar_chart_df.sort_values(by = metric, ascending = is_barchart_sorted)

    metric_name = metric_display_names[metric]
    dimension_name = dimension_labels[dimension]

    title = f"{metric_name} {dimension_name}"

    height = max(400, bar_chart_df.shape[0] * 20)
    bar_chart_figure = px.bar(data_frame = bar_chart_df, x = metric, y = dimension, orientation = 'h', height = height)
   
    bar_chart_figure.update_layout(
        yaxis={"tickformat": ",.0f"}, 
        paper_bgcolor="#FDF0D5",
        plot_bgcolor = "#748CAB",   
        title_text = title,
        title_font_color = "#1D2D44",
        title_font_family = "Arial",
        title_font_size = 18,
        title_font_weight = 600, 
        margin=dict(l=25, r=25, b=25, t=65, pad = 5)
        )
    
    bar_chart_figure.update_traces(width = 0.5, marker_color = "#3E5C76", marker_line_color = "#1D2D44", marker_line_width = 1.5, opacity = 1)
    bar_chart_figure.update_xaxes(tickformat =  metric_config[metric]["format"], gridcolor = "#F0EBD8", title_text = metric, fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", tickfont_weight = 600, tickfont_family = "Arial", tickfont_size = 9,  tickfont_color = "#0D1321", ticklabelstandoff = 5)
    bar_chart_figure.update_yaxes(fixedrange = True, tickfont_color = "#0D1321",  showgrid = False, tickfont_family = "Arial", tickfont_weight = 600, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", ticklabelstandoff = 5)    


    return bar_chart_figure, {
        "position": "relative",
        "max-width":"20px",
        "max-height":"20px",
        "width": "auto",
        "height": "auto",
        "margin-left": "2px",
        "transform": "rotate(0deg)" if is_barchart_sorted else "rotate(180deg)",
        "transition": "transform 0.2s ease"
    }

@app.callback(Output("scatterplot_graph", "figure"), [Input("scatterplot_metric_x_dropdown", "value"), Input("scatterplot_metric_y_dropdown", "value"), Input("scatterplot_dimension_dropdown", "value")])
def update_scatterplot(metric_x, metric_y, dimension):

    scatterplot_df = data_frame.copy()
    scatterplot_df["metric_x"] = scatterplot_df[metric_x]
    scatterplot_df["metric_y"] = scatterplot_df[metric_y]
    if dimension == "None":
        scatterplot_df = scatterplot_df[["metric_x", "metric_y"]]
    else:
        scatterplot_df = scatterplot_df[["metric_x", "metric_y", dimension]]

    title = f"Relationship between {metric_x} and {metric_y}"
 
    if dimension == "None":
        scatterplot_fig = px.scatter(data_frame = scatterplot_df, x = "metric_x", y = "metric_y", trendline="ols")
    else:
        scatterplot_fig = px.scatter(data_frame = scatterplot_df, x = "metric_x", y = "metric_y", trendline="ols", color = dimension, color_discrete_sequence = line_chart_color_palette)
   
    scatterplot_fig.update_layout(
        yaxis={"tickformat": ",.0f"}, 
        paper_bgcolor="#FDF0D5",
        plot_bgcolor = "#748CAB",   
        title_text = title,
        title_font_color = "#1D2D44",
        title_font_family = "Arial",
        title_font_size = 18,
        title_font_weight = 600, 
        margin=dict(l=25, r=25, b=25, t=65, pad = 5)
        )
    
    if dimension == "None":
        scatterplot_fig.update_traces(marker_size = 4, marker_color = "#3E5C76", marker_line_width = 0, opacity = 1)
    else:
        scatterplot_fig.update_traces(marker_size = 5, marker_line_width = 0, opacity = 1)

    scatterplot_fig.update_xaxes(
        zeroline = False, tickformat =  metric_config[metric_x]["format"], gridcolor = "#F0EBD8", title_text = metric_x, 
        fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", tickfont_weight = 600, 
        tickfont_family = "Arial", tickfont_size = 9,  tickfont_color = "#0D1321", ticklabelstandoff = 5)
    scatterplot_fig.update_yaxes(
        zeroline = False, tickformat =  metric_config[metric_y]["format"], gridcolor = "#F0EBD8", title_text = metric_y, 
        fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", tickfont_weight = 600,
        tickfont_family = "Arial", tickfont_size = 9, tickfont_color = "#0D1321", ticklabelstandoff = 5)    


    return scatterplot_fig





def build_sim_pareto(pareto_chart_df, pareto_dimension):
    pareto_chart_df = pareto_chart_df.groupby(pareto_dimension)["what_if_profit"].sum().reset_index()
    pareto_chart_df = pareto_chart_df.loc[pareto_chart_df["what_if_profit"] > 0]
    pareto_chart_df = pareto_chart_df.sort_values(by = "what_if_profit", ascending = False)
    pareto_chart_df["percentage"] = (pareto_chart_df["what_if_profit"] * 100) / pareto_chart_df["what_if_profit"].sum()
    pareto_chart_df["cumulative_percentage"] = pareto_chart_df["percentage"].cumsum()
    
    dimension_name = dimension_labels[pareto_dimension]
    title = f"Profit {dimension_name}"

    pareto_chart_bar_data = Bar(x = pareto_chart_df[pareto_dimension], y = pareto_chart_df["what_if_profit"])
    pareto_chart_scatter_data = Scatter(x = pareto_chart_df[pareto_dimension], y = pareto_chart_df["cumulative_percentage"],  yaxis = "y2", mode = "lines+markers")
    pareto_chart_figure = Figure(data = [pareto_chart_bar_data, pareto_chart_scatter_data])
    pareto_chart_figure.update_layout(
        showlegend = False,
        height = 450, 
        paper_bgcolor="#FDF0D5",
        plot_bgcolor = "#748CAB",   
        title_text = title,
        title_font_color = "#1D2D44",
        title_font_family = "Arial",
        title_font_size = 18,
        title_font_weight = 600, 
        margin = dict(l=25, r=25, b=25, t=65, pad = 5),
        yaxis2 = dict(
            side="right",
            overlaying="y",
            range=[0, 105],    
            ticksuffix="%",
            showgrid=False,
            fixedrange=True,
            tickfont_family="Arial",
            tickfont_weight=600,
            tickfont_size=10,
            tickfont_color="#0D1321",
            ticklabelstandoff=5,
            title_text = "Cumulative Percentage (%)",
            title_font_family="Arial",
            title_font_weight=600,
            title_font_color="#0D1321"
        ),
        yaxis = dict(
            tickformat=",.0f",
            fixedrange=True,
            tickfont_color="#0D1321",
            tickfont_family="Arial",
            tickfont_weight=600,
            tickfont_size=10,
            gridcolor = "#F0EBD8",
            title_text = "Profit",
            title_font_family="Arial",
            title_font_weight=600,
            title_font_color="#0D1321",
            ticklabelstandoff=5
        ),
        xaxis = dict(
            tickformat =  ",.0f", 
            showgrid=False,
            title_text = pareto_dimension, 
            fixedrange = True, 
            title_font_family = "Arial", 
            title_font_weight = 600, 
            title_font_color = "#0D1321", 
            tickfont_weight = 600, 
            tickfont_family = "Arial", 
            tickfont_size = 9,  
            tickfont_color = "#0D1321", 
            ticklabelstandoff = 5,
            automargin=True,
            range=[-0.6, len(pareto_chart_df) - 0.4]
        )
    )


    pareto_chart_figure.update_traces(width = 0.5, marker_color = "#3E5C76", marker_line_color = "#1D2D44", marker_line_width = 1.5, opacity = 1, selector=dict(type='bar'))
    pareto_chart_figure.update_traces(mode="markers+lines", marker_line_width = 1.5, line_width = 2, marker_size = 5, marker_color = "#E07E50", marker_line_color = "#1D2D44", selector=dict(type='scatter'))

    return pareto_chart_figure

def build_sim_profit_linechart(line_chart_df, time_granularity, dimension):
    if time_granularity == "month":
        line_chart_df["date"] = line_chart_df["order_date"].dt.to_period("M").dt.to_timestamp()
        line_chart_df["label"] = line_chart_df["date"].dt.strftime("%b %Y")

    elif time_granularity == "quarter":
        line_chart_df["date"] = line_chart_df["order_date"].dt.to_period("Q").dt.to_timestamp()
        line_chart_df["label"] = (
            line_chart_df["date"].dt.year.astype(str) +
            " Q" +
            line_chart_df["date"].dt.quarter.astype(str)
        )

    elif time_granularity == "year":
        line_chart_df["date"] = line_chart_df["order_date"].dt.to_period("Y").dt.to_timestamp()
        line_chart_df["label"] = line_chart_df["date"].dt.year.astype(str)


    group_keys = [dimension, "date", "label"] if dimension != "None" else ["date", "label"]
    line_chart_df = line_chart_df.groupby(group_keys)["what_if_profit"].sum().reset_index()
    line_chart_df = line_chart_df.sort_values(by="date")


    time_name = time_granularity_labels[time_granularity]
    dimension_name = dimension_labels[dimension]

    title = f"{time_name} profit"
    if dimension_name:
        title += f" {dimension_name}"


    line_chart_fig = px.line(data_frame = line_chart_df, x = "date", y = "what_if_profit", color = (None if dimension == "None" else dimension), color_discrete_sequence = line_chart_color_palette)
    line_chart_fig.update_layout(
                yaxis={"tickformat": ",.0f"}, 
                paper_bgcolor="#FDF0D5",
                plot_bgcolor = "#748CAB",   
                title_text = title,
                title_font_color = "#1D2D44",
                title_font_family = "Arial",
                title_font_size = 18,
                title_font_weight = 600,
                title_y = 0.95, 
                margin=dict(l=25, r=25, b=25, t=65, pad = 5))

    unique_dates_df = line_chart_df[["date", "label"]].drop_duplicates().sort_values(by="date")
    tickvals_x = unique_dates_df["date"][::3] if time_granularity == "month" else unique_dates_df["date"]
    ticktext_x = unique_dates_df["label"][::3]  if time_granularity == "month" else unique_dates_df["label"]
    

    line_chart_fig.update_traces(mode="markers+lines", marker_line_width = 1.5, line_width = 2, marker_size = 5)
    line_chart_fig.update_xaxes(range = [unique_dates_df["date"].min(), unique_dates_df["date"].max()],tickvals = tickvals_x, ticktext = ticktext_x, tickangle = 45, showgrid = False, title_text = "Date", fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", title_font_size = 12, tickfont_weight = 600, tickfont_family = "Arial", tickfont_size = 9,  tickfont_color = "#0D1321", ticklabelstandoff = 5)
    line_chart_fig.update_yaxes(autorange = True, tickformat =  ",.0f", zeroline = False, fixedrange = True, tickfont_color = "#0D1321", tickfont_family = "Arial", tickfont_weight = 600, tickfont_size = 10, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", title_font_size = 12, title_text = "Profit", gridcolor = "#F0EBD8", ticklabelstandoff = 5)    

    return line_chart_fig

def build_sim_profit_barchart(bar_chart_df, barchart_dimension):
    global is_sim_profit_barchart_sorted
    if ctx.triggered_id == "sim_profit_barchart_sort_button": 
        is_sim_profit_barchart_sorted = not is_sim_profit_barchart_sorted


    bar_chart_df = bar_chart_df.groupby(barchart_dimension)["what_if_profit"].sum().reset_index()
    bar_chart_df = bar_chart_df.sort_values(by = "what_if_profit", ascending = is_sim_profit_barchart_sorted)
  

    dimension_name = dimension_labels[barchart_dimension]
    title = f"Profit {dimension_name}"

    height = max(450, bar_chart_df.shape[0] * 20)
    bar_chart_figure = px.bar(data_frame = bar_chart_df, x = "what_if_profit", y = barchart_dimension, orientation = 'h', height = height)
    bar_chart_figure.update_layout(
        yaxis={"tickformat": ",.0f"}, 
        paper_bgcolor="#FDF0D5",
        plot_bgcolor = "#748CAB",   
        title_text = title,
        title_font_color = "#1D2D44",
        title_font_family = "Arial",
        title_font_size = 18,
        title_font_weight = 600, 
        margin=dict(l=25, r=25, b=25, t=65, pad = 5)
        )
    
    bar_chart_figure.update_traces(width = 0.5, marker_color = "#3E5C76", marker_line_color = "#1D2D44", marker_line_width = 1.5, opacity = 1)
    bar_chart_figure.update_xaxes(tickformat =  ",.0f", gridcolor = "#F0EBD8", title_text = "Profit", fixedrange = True, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", tickfont_weight = 600, tickfont_family = "Arial", tickfont_size = 9,  tickfont_color = "#0D1321", ticklabelstandoff = 5)
    bar_chart_figure.update_yaxes(fixedrange = True, tickfont_color = "#0D1321",  showgrid = False, tickfont_family = "Arial", tickfont_weight = 600, tickfont_size = 10, title_font_family = "Arial", title_font_weight = 600, title_font_color = "#0D1321", ticklabelstandoff = 5)    


    sort_arrow_style = {
        "position": "relative",
        "max-width":"15px",
        "max-height":"15px",
        "width": "auto",
        "height": "auto",
        "margin-left": "2px",
        "transform": "rotate(0deg)" if is_sim_profit_barchart_sorted else "rotate(180deg)",
        "transition": "transform 0.2s ease"
    }

    return bar_chart_figure, sort_arrow_style

is_sim_profit_barchart_sorted = True
@app.callback([Output("sim_profit_linechart_graph", "figure"), Output("sim_profit_barchart_graph", "figure"), Output("sim_profit_barchart_sort_icon_arrow", "style"), Output("sim_profit_pareto_graph", "figure")], 
              [Input("sim_profit_time_granularity_dropdown", "value"), Input("sim_profit_filter_dropdown", "value"), 
               Input("sim_prod_cost_slider", "value"), Input("sim_discount_slider", "value"), Input("sim_ship_cost_slider", "value"), 
               Input("sim_profit_barchart_sort_button", "n_clicks"), Input("sim_profit_barchart_dimension_dropdown", "value"), Input("sim_profit_pareto_dimension_dropdown", "value")])
def update_what_if_analysis(time_granularity, dimension, cost_change, discount_change, shippingcost_change, sort_button_clicks, barchart_dimension, pareto_dimension):
    df = data_frame.copy()
    
    cost_factor = 1 + (cost_change / 100)
    ship_factor = 1 + (shippingcost_change / 100)
    discount_val = discount_change / 100

    df["new_discount"] = (df["discount"] + discount_val).clip(0, 1)
    df["sales"] = df["original_price"] * (1 - df["new_discount"])
    df["what_if_profit"] = df["sales"] - ((df["product_cost"] * cost_factor) + (df["shipping_cost"] * ship_factor))
    df["order_date"] = pd.to_datetime(df["order_date"])

    pareto_chart_df = df.copy()
    line_chart_df = df.copy()
    bar_chart_df = df.copy()
    

    if ctx.triggered_id == "sim_profit_barchart_sort_button" or ctx.triggered_id == "sim_profit_barchart_dimension_dropdown":
        barchart_fig, sort_arrow_style = build_sim_profit_barchart(bar_chart_df, barchart_dimension)
        return no_update, barchart_fig, sort_arrow_style, no_update
    elif ctx.triggered_id == "sim_profit_time_granularity_dropdown" or ctx.triggered_id == "sim_profit_filter_dropdown":
        linechart_fig = build_sim_profit_linechart(line_chart_df, time_granularity, dimension)
        return linechart_fig, no_update, no_update, no_update
    elif ctx.triggered_id == "sim_profit_pareto_dimension_dropdown":
        pareto_fig = build_sim_pareto(pareto_chart_df, pareto_dimension)
        return no_update, no_update, no_update, pareto_fig
    else:
        linechart_fig = build_sim_profit_linechart(line_chart_df, time_granularity, dimension)
        barchart_fig, sort_arrow_style = build_sim_profit_barchart(bar_chart_df, barchart_dimension)
        pareto_fig = build_sim_pareto(pareto_chart_df, pareto_dimension)

        return linechart_fig, barchart_fig, sort_arrow_style, pareto_fig




app.run(host = "127.0.0.1", port = 8050, debug = False)
```
